```python
from pathlib import Path
import pandas as pd

# Notebook is running from: marketing_campaign/notebooks
ROOT = Path.cwd().parent  # go up from notebooks → marketing_campaign

DATA_DIR = ROOT / "backend" / "data" / "bank_data"

print("ROOT:", ROOT)
print("DATA_DIR:", DATA_DIR)
print("Exists:", DATA_DIR.exists())
print("Files:", list(DATA_DIR.iterdir()))

# ✅ change this filename to whatever is inside your data/ folder
csv_path = DATA_DIR / "bank-additional-full.csv"

df = pd.read_csv(csv_path, sep=";")  # UCI bank marketing uses ';'
df.columns = [c.strip().lower().replace(".", "_") for c in df.columns]

# target -> 0/1
if "y" in df.columns:
    df["y"] = df["y"].map({"yes": 1, "no": 0}).astype("int64")
else:
    raise ValueError("Target column 'y' not found. Check the CSV you downloaded.")

print("shape:", df.shape)
print("target counts:\n", df["y"].value_counts())
print("top missing %:\n", (df.isna().mean().sort_values(ascending=False).head(10) * 100).round(2))

# Features/target
X = df.drop(columns=["y"])
y = df["y"]

# One-hot encode categoricals
cat_cols = X.select_dtypes(include=["object"]).columns.tolist()
X_enc = pd.get_dummies(X, columns=cat_cols, drop_first=False)

# Final table
model_df = X_enc.copy()
model_df["y"] = y.values

# Save
out_path = DATA_DIR / "model_ready.parquet"
model_df.to_parquet(out_path, index=False)

print("saved:", out_path)
print("final shape:", model_df.shape)
```
